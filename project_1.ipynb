```python
def iris_classify_v3(length, width, setosa_length=2.0,
                  setosa_width=0.7, versicolor_length=5.1,
                  versicolor_width=1.8):

    if length <= setosa_length and width <= setosa_width:
        return 0

    elif length <= versicolor_length and width <= versicolor_width:
        return 1

    else:
        return 2
```

```python
best_accuracy = 0
best_thresholds = None

for sl in np.arange(1.5, 2.6, 0.1):
    for sw in np.arange(0.4, 0.9, 0.05):
        for vl in np.arange(4.5, 5.6, 0.1):
            for vw in np.arange(1.5, 2.1, 0.05):

                predictions = np.array([
                    iris_classify(
                        float(row[2]),
                        float(row[3]),
                        sl, sw, vl, vw
                    )
                    for row in x_train
                ])

                predicted_species = target_names[predictions]
                actual = x_train[:, 4]

                accuracy = np.mean(actual == predicted_species) * 100

                if accuracy > best_accuracy:
                    best_accuracy = accuracy
                    best_thresholds = (sl, sw, vl, vw)

print("Best accuracy:", best_accuracy)
print("Best thresholds:", best_thresholds)
```

Training stage -
X_train + y_train
        |
Model learns patterns


Prediction stage -
X_test
      |
Model generates predictions


Evaluation stage -
Compare predictions against y_test


The most important distinction:
- model.fit(X_train, y_train) — trains the model.
- model.predict(X_test) — predicts unseen observations.
- Compare predictions with y_test — evaluates performance.
Never use y_test to train the model. It must remain separate so we can fairly evaluate the model.
